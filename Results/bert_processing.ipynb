```python
import pandas as pd

# Loading
df = pd.read_csv("camera_2013_2022_textcleaned.csv")  # or your actual file path
```

```python
# Checking whether the cleaned text may be longer than 512 tokens
df["word_count"] = df["text_clean"].apply(lambda x: len(str(x).split()))

# Inspect
print("Average length:", df["word_count"].mean())
print("Max length:", df["word_count"].max())

# To see which article is longest:
longest_text = df.loc[df["word_count"].idxmax(), "text_clean"]
print("Longest text sample:\n", longest_text[:500], "...")

#conclusion: we need to do something before running BERT otherwise we lose the rest of the speech
```
